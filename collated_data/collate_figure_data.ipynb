```python
import pathlib as pl
import pandas as pd
import numpy as np
```

```python
collated_data_path = pl.Path.cwd()
root_path = collated_data_path.parent

all_csv = root_path.glob("fig_*/code/*.csv")
collated_plot_data_file = collated_data_path / 'collated_plot_data.xlsx'

with pd.ExcelWriter(collated_plot_data_file, engine='openpyxl') as writer:

    for pth in all_csv:
        df = pd.read_csv(pth)
        fig_name = pth.stem
        df.to_excel(writer, sheet_name=fig_name, index=False)
```
